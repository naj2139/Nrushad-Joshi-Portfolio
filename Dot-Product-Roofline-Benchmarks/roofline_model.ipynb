```python
import numpy as np
import matplotlib.pyplot as plt

I = np.logspace(-1, 2, 500)
roof = np.minimum(200, 30 * I)


perf = [2.120, 7.616, 31.277, 0.013, 14.840]

plt.plot(I, roof, color="black")
plt.axvline(0.25, color="gray", linestyle="--")
plt.scatter([0.25] * 5, perf)

plt.text(0.28, 2.115, "C1")
plt.text(0.28, 7.364, "C2")
plt.text(0.28, 45.710, "C3")
plt.text(0.28, 0.0131, "C4")
plt.text(0.28, 17.456, "C5")

plt.xscale("log")
plt.yscale("log")
plt.xlabel("Arithmetic intensity (FLOP/byte)")
plt.ylabel("Performance (GFLOP/s)")
plt.title("N = 1,000,000")
plt.savefig("/content/drive/MyDrive/Roofline_1M.png")
```
